# **Experiment 6 : Classification of Iris Flower Species using Support Vector Machine (SVM)**

## **Objective**

The primary objective of this notebook is to apply **Support Vector Machine (SVM)**, a supervised machine learning algorithm, to the **Iris dataset** in order to classify iris flowers into their respective species. The aim is to build a predictive model that can accurately distinguish between *Setosa*, *Versicolor*, and *Virginica* using the dataset’s features such as petal length, petal width, sepal length, and sepal width.


## **Problem Overview**

The **Iris dataset** is one of the most widely used datasets for beginner-level classification problems. It contains 150 samples with four features and three target classes. The problem is essentially a **multi-class classification task**, where the challenge is to map feature values to the correct species. Traditional algorithms like logistic regression or decision trees can handle this dataset, but using **SVM provides a more robust approach** for achieving higher accuracy and understanding the concept of hyperplanes and decision boundaries.

## **Introduction**

Classification plays a key role in machine learning, especially in problems where the goal is to assign labels to input data. The Iris dataset, introduced by Ronald A. Fisher, has become a benchmark for testing classification algorithms. In this notebook, the Support Vector Machine algorithm is employed because of its effectiveness in creating optimal decision boundaries.

SVM works by finding a hyperplane that best separates data points of different classes. For linearly separable data, SVM identifies the hyperplane with maximum margin between the classes, whereas for non-linear data, it uses kernel tricks to map inputs to higher-dimensional spaces for better separation. This makes SVM particularly effective in solving classification problems like the Iris dataset.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
```

```python
df=pd.read_csv('IRIS.csv')
df.head()
```

```python
df.species.replace(('Iris-setosa', 'Iris-versicolor', 'Iris-virginica'), (1,2,3), inplace=True)
```

```python
df.info()
```

```python
plt.figure(figsize = (10, 5))
df.boxplot(color='red')
```

```python
df1=df[df.species==1]
df2=df[df.species==2]
df3=df[df.species==3]
```

```python
plt.figure(figsize = (10, 5))
plt.scatter(x=df1['sepal_length'],y=df1['sepal_width'], color='green')
plt.scatter(x=df2['sepal_length'],y=df2['sepal_width'], color='blue')
```

```python
plt.figure(figsize = (10, 5))
plt.scatter(x=df1['sepal_length'],y=df1['sepal_width'], color='green')
plt.scatter(x=df3['sepal_length'],y=df3['sepal_width'], color='red')
```

```python
x=df.drop(['species'],axis='columns')
x.head()
```

```python
y=df.species
y.head()
```

```python
x_train,x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state = 2)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
model = SVC()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test )
```
